```python
import pandas as pd
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
```

```python
df = pd.read_csv('data/housing-2.csv')
df.head()
```

```python
X = df[['year', 'price']]
sns.scatterplot(x=X.year, y=X.price)
```

```python
gm = GaussianMixture(n_components=3, random_state=42)

gm.fit(X)

gm_pred = gm.predict(X)

sns.scatterplot(x=X.year, y=X.price, c=gm_pred)
```

```python
kmeans = KMeans(n_clusters=3)
kmeans.fit(X)
kmeans_pred = kmeans.predict(X)
sns.scatterplot(x=X.year, y=X.price, c=kmeans_pred)
```
