```python
# пропуски данных
print(data.isnull().sum())
```

```python
# удаление признаков с большим количеством пропусков
cols_to_drop = ['region_2', 'taster_twitter_handle']
data = data.drop(cols_to_drop, axis=1)

# удаление строк с пропусками
data = data.dropna(subset=['variety'])

# заполнение пропусков
data['country'] = data['country'].fillna('Unknown')
data['province'] = data['province'].fillna('Unknown')
data['region_1'] = data['region_1'].fillna('Unknown')
data['designation'] = data['designation'].fillna('Unknown')
data['taster_name'] = data['taster_name'].fillna('Unknown')

# заполнение цены медианным значением
data['price'] = data['price'].fillna(data['price'].median())
```

```python
# пропуски данных
print(data.isnull().sum())
```
